```python
import zipfile
import pandas as pd
import pyreadstat
import os
```

```python

# Caminho para o arquivo ZIP
caminho_zip = r"C:\Users\ribei\OneDrive\PDM\ciencia-de-dados\conversao_arquivo_icbeu\15012025_0700.zip"

# Nome do arquivo .bkp dentro do zip
nome_arquivo_bkp = "scl.bkp"  # Substitua pelo nome exato do arquivo .bkp dentro do zip
```

```python

# Lendo o arquivo .bkp dentro do zip e exibindo seu conteúdo
def ler_arquivo_bkp_no_zip(caminho_zip, nome_arquivo_bkp):
    try:
        # Abre o arquivo ZIP
        with zipfile.ZipFile(caminho_zip, 'r') as zip_ref:
            # Lista os arquivos no ZIP
            arquivos_no_zip = zip_ref.namelist()
            print("Arquivos no ZIP:", arquivos_no_zip)
            
            # Verifica se o arquivo .bkp existe no ZIP
            if nome_arquivo_bkp not in arquivos_no_zip:
                print(f"O arquivo {nome_arquivo_bkp} não foi encontrado no ZIP.")
                return
            
            # Lê o conteúdo do arquivo .bkp
            with zip_ref.open(nome_arquivo_bkp) as arquivo_bkp:
                conteudo = arquivo_bkp.read()
            
            # Exibe uma parte do conteúdo
            print("Conteúdo inicial do arquivo (.bkp):")
            print(conteudo[:500])  # Exibe os primeiros 500 bytes do arquivo como exemplo
            return conteudo
    except Exception as e:
        print(f"Erro ao processar o arquivo: {e}")
        return None
```

```python
# Executa a função
conteudo_bkp = ler_arquivo_bkp_no_zip(caminho_zip, nome_arquivo_bkp)
```
