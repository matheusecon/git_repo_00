```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut
```

```python
# Função para obter latitude e longitude a partir de endereço ou CEP
def get_lat_long(query):
    geolocator = Nominatim(user_agent="nome_do_seu_projeto_geocoding")
    try:
        location = geolocator.geocode(query, timeout=10)
        if location:
            return (location.latitude, location.longitude)
        else:
            return (None, None)
    except GeocoderTimedOut:
        return (None, None)
```

```python
data = {
    'endereco': ['Rua Exemplo 123', None, 'Av Paulista 1000', None],
    'cep': [None, '01000-000', None, '01310-200']
}
df = pd.DataFrame(data)
```

```python
display(df)
```

```python
def preencher_lat_long(row):
    if pd.notnull(row['endereco']):
        return get_lat_long(row['endereco'])
    elif pd.notnull(row['cep']):
        return get_lat_long(row['cep'])
    else:
        return (None, None)
```

```python

# Aplicando a função ao DataFrame para obter latitude e longitude
df[['latitude', 'longitude']] = df.apply(preencher_lat_long, axis=1, result_type="expand")
display(df)
```
